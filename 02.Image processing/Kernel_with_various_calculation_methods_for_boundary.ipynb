```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

# Load a sample image

```python
image_path = './sample2.PNG'
image = cv2.imread(image_path)
img_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
```

# Define blur kernel


```python
kernel = np.array([[1, 1, 1],
                   [1, 1, 1],
                   [1, 1, 1]])/9
```

# Kernel with various calculation methods for boundary

```python
# full
blurred_full = cv2.filter2D(img_rgb, -1, kernel, borderType=cv2.BORDER_REPLICATE)  # input img, output img depth(bit), kernel, method for boundary


# same
blurred_same = cv2.filter2D(img_rgb, -1, kernel, borderType=cv2.BORDER_REFLECT)


# valid
blurred_valid = cv2.filter2D(img_rgb, -1, kernel, borderType=cv2.BORDER_CONSTANT)
```

```python
# 시각화
fig, axs = plt.subplots(1, 4, figsize=(20, 5))

axs[0].imshow(img_rgb, cmap='gray')
axs[0].set_title('Original Image')
axs[0].axis('off')

axs[1].imshow(blurred_full, cmap='gray')
axs[1].set_title('Full - Replicate Border')
axs[1].axis('off')

axs[2].imshow(blurred_same, cmap='gray')
axs[2].set_title('Same - Reflect Border')
axs[2].axis('off')

axs[3].imshow(blurred_valid, cmap='gray')
axs[3].set_title('Valid - Constant (Pad 0)')
axs[3].axis('off')

plt.show()
```

△ if you look it closely, 'fill with 0' method image has a black boundary
